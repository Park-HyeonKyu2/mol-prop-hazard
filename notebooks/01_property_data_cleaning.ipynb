```python
tmp = pd.read_csv("./data/prop/crc_mp_c.csv")
tbp = pd.read_csv("./data/prop/crc_bp_c.csv")
vp = pd.read_csv("./data/prop/crc_vp_kpa_25c_filtered.csv")

print("MP:", tmp.shape, tmp.columns.tolist())
print("BP:", tbp.shape, tbp.columns.tolist())
print("VP:", vp.shape, vp.columns.tolist())
```

```python
from rdkit import Chem

tmp["mol"] = tmp["SMILES"].apply(Chem.MolFromSmiles)
success_rate = tmp["mol"].notna().mean()
print(f"tmp 파싱 성공률: {success_rate*100:.1f}%")

tbp["mol"] = tbp["SMILES"].apply(Chem.MolFromSmiles)
success_rate = tbp["mol"].notna().mean()
print(f"tbp 파싱 성공률: {success_rate*100:.1f}%")

vp["mol"] = vp["SMILES"].apply(Chem.MolFromSmiles)
success_rate = vp["mol"].notna().mean()
print(f"vp 파싱 성공률: {success_rate*100:.1f}%")

tc["mol"] = tc["SMILES"].apply(Chem.MolFromSmiles)
success_rate = tc["mol"].notna().mean()
print(f"tc 파싱 성공률: {success_rate*100:.1f}%")

vc["mol"] = vc["SMILES"].apply(Chem.MolFromSmiles)
success_rate = vc["mol"].notna().mean()
print(f"vc 파싱 성공률: {success_rate*100:.1f}%")

pc["mol"] = pc["SMILES"].apply(Chem.MolFromSmiles)
success_rate = pc["mol"].notna().mean()
print(f"pc 파싱 성공률: {success_rate*100:.1f}%")
```

```python
from rdkit import Chem

# SMILES를 미리 파싱해둔 mol 객체를 받아서 InChIKey를 계산하는 함수
def get_inchikey(mol):
    if mol is None:
        return None
    else:
        inchikey = Chem.MolToInchiKey(mol)
        return inchikey


# ---------- tmp (녹는점) ----------
inchikey_list = []
for mol in tmp["mol"]:
    key = get_inchikey(mol)
    inchikey_list.append(key)
tmp["InChIKey"] = inchikey_list
print("tmp 완료:", tmp["InChIKey"].notna().sum(), "/", len(tmp))


# ---------- tbp (끓는점) ----------
inchikey_list = []
for mol in tbp["mol"]:
    key = get_inchikey(mol)
    inchikey_list.append(key)
tbp["InChIKey"] = inchikey_list
print("tbp 완료:", tbp["InChIKey"].notna().sum(), "/", len(tbp))


# ---------- vp (증기압) ----------
inchikey_list = []
for mol in vp["mol"]:
    key = get_inchikey(mol)
    inchikey_list.append(key)
vp["InChIKey"] = inchikey_list
print("vp 완료:", vp["InChIKey"].notna().sum(), "/", len(vp))


# ---------- tc (임계온도) ----------
inchikey_list = []
for mol in tc["mol"]:
    key = get_inchikey(mol)
    inchikey_list.append(key)
tc["InChIKey"] = inchikey_list
print("tc 완료:", tc["InChIKey"].notna().sum(), "/", len(tc))


# ---------- vc (임계부피) ----------
inchikey_list = []
for mol in vc["mol"]:
    key = get_inchikey(mol)
    inchikey_list.append(key)
vc["InChIKey"] = inchikey_list
print("vc 완료:", vc["InChIKey"].notna().sum(), "/", len(vc))


# ---------- pc (임계압력) ----------
inchikey_list = []
for mol in pc["mol"]:
    key = get_inchikey(mol)
    inchikey_list.append(key)
pc["InChIKey"] = inchikey_list
print("pc 완료:", pc["InChIKey"].notna().sum(), "/", len(pc))
```

```python
def check_quality(df, smiles_col, value_col):
    dec_mask = df[value_col].astype(str).str.contains("dec", na=False)
    metal_mask = df[smiles_col].str.contains(r"\[(Ca|Na|K|Mg|Fe|Zn|Al|Li)\+", regex=True, na=False)
    print(f"'dec' 포함: {dec_mask.sum()}개")
    print(f"금속 양이온 포함: {metal_mask.sum()}개")
    print(f"합쳐서 제외 후보: {(dec_mask | metal_mask).sum()}개 / 전체 {len(df)}개")
    return dec_mask, metal_mask

print("=== tmp (녹는점) ===")
check_quality(tmp, "SMILES", "tmp/ºC")

print("\n=== tbp (끓는점) ===")
check_quality(tbp, "SMILES", "tbp/ºC")

print("\n=== vp (증기압) ===")
check_quality(vp, "SMILES", "vp/kPa (25 °C)")

print("\n=== tc (임계온도) ===")
check_quality(tc, "SMILES", "Tc/K")

print("\n=== vc (임계부피) ===")
check_quality(vc, "SMILES", "Vc/cm3 mol-1")

print("\n=== pc (임계압력) ===")
check_quality(pc, "SMILES", "Pc/MPa")
```

```python
# ---- tmp ----
tmp_dec_mask = tmp["tmp/ºC"].astype(str).str.contains("dec", na=False)
tmp_metal_mask = tmp["SMILES"].str.contains(r"\[(Ca|Na|K|Mg|Fe|Zn|Al|Li)\+", regex=True, na=False)
tmp_subl_mask = tmp["tmp/ºC"].astype(str).str.contains("subl", na=False)
tmp_range_mask = tmp["tmp/ºC"].astype(str).str.contains(r"[<>]|(\d\s*-\s*\d)", regex=True, na=False)

print("tmp dec:", tmp_dec_mask.sum(), "/ metal:", tmp_metal_mask.sum(), "/ subl:", tmp_subl_mask.sum(), "/ range:", tmp_range_mask.sum())

# ---- tbp ----
tbp_dec_mask = tbp["tbp/ºC"].astype(str).str.contains("dec", na=False)
tbp_metal_mask = tbp["SMILES"].str.contains(r"\[(Ca|Na|K|Mg|Fe|Zn|Al|Li)\+", regex=True, na=False)
tbp_subl_mask = tbp["tbp/ºC"].astype(str).str.contains("subl", na=False)
tbp_range_mask = tbp["tbp/ºC"].astype(str).str.contains(r"[<>]|(\d\s*-\s*\d)", regex=True, na=False)

print("tbp dec:", tbp_dec_mask.sum(), "/ metal:", tbp_metal_mask.sum(), "/ subl:", tbp_subl_mask.sum(), "/ range:", tbp_range_mask.sum())
```

```python
total_before = len(tmp)
tmp_clean = tmp[~(tmp_dec_mask | tmp_metal_mask | tmp_subl_mask | tmp_range_mask)]
print(f"tmp: {total_before}개 -> {len(tmp_clean)}개 ({total_before - len(tmp_clean)}개 제거)")

total_before = len(tbp)
tbp_clean = tbp[~(tbp_dec_mask | tbp_metal_mask | tbp_subl_mask | tbp_range_mask)]
print(f"tbp: {total_before}개 -> {len(tbp_clean)}개 ({total_before - len(tbp_clean)}개 제거)")
```

```python
from rdkit import Chem

def get_inchikey(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None
    return Chem.MolToInchiKey(mol)

# 정제된 tmp, tbp
tmp_clean["InChIKey"] = tmp_clean["SMILES"].apply(get_inchikey)
tbp_clean["InChIKey"] = tbp_clean["SMILES"].apply(get_inchikey)

tmp_clean.drop(columns=["mol"]).to_csv("./data/processed/crc_mp_c.csv", index=False)
tbp_clean.drop(columns=["mol"]).to_csv("./data/processed/crc_bp_c.csv", index=False)

# 정제 필요 없는 vp, tc, vc(raw), pc — InChIKey는 이미 앞에서 만들어둔 mol 컬럼 재사용
vp.drop(columns=["mol"]).to_csv("./data/processed/crc_vp_kpa_25c_filtered.csv", index=False)
tc.drop(columns=["mol"]).to_csv("./data/processed/crc_tc_k.csv", index=False)
vc.drop(columns=["mol"]).to_csv("./data/processed/crc_vc_cm3mol_raw.csv", index=False)
pc.drop(columns=["mol"]).to_csv("./data/processed/crc_pc_mpa.csv", index=False)
```
