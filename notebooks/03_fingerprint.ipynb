```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import rdFingerprintGenerator

# Morgan fingerprint(ECFP) 생성기: radius=2, 길이 2048비트
morgan_gen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)

def get_fingerprint(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None
    fp = morgan_gen.GetFingerprint(mol)
    return np.array(fp)
```

```python
# ---------- mp (녹는점) ----------
mp = pd.read_csv("./data/processed/crc_mp_c.csv")

fp_list = []
for smiles in mp["SMILES"]:
    fp = get_fingerprint(smiles)
    fp_list.append(fp)

mp_fp = np.array(fp_list)
print("mp fingerprint shape:", mp_fp.shape)

np.save("./data/processed/crc_mp_c_fingerprint.npy", mp_fp)
```

```python
# ---------- bp (끓는점) ----------
bp = pd.read_csv("./data/processed/crc_bp_c.csv")

fp_list = []
for smiles in bp["SMILES"]:
    fp = get_fingerprint(smiles)
    fp_list.append(fp)

bp_fp = np.array(fp_list)
print("bp fingerprint shape:", bp_fp.shape)

np.save("./data/processed/crc_bp_c_fingerprint.npy", bp_fp)
```
